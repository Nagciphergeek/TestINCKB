```python
# Environment setup & imports

!python --version

import os
import pathlib
import shutil
import subprocess
import json

import nbformat
from nbclient import NotebookClient

print("Environment setup cells loaded")
```

```python
# Inspect incident2kb folder (snapshot & manifest)

from pathlib import Path
import hashlib

root = Path.cwd()
incidents_dir = root
manifest = []
for path in sorted(incidents_dir.rglob('*')):
    if path.is_file():
        sha256 = hashlib.sha256(path.read_bytes()).hexdigest()
        manifest.append({
            'path': str(path.relative_to(root)),
            'size': path.stat().st_size,
            'sha256': sha256,
        })

manifest_path = root / 'manifest_before.json'
manifest_path.write_text(json.dumps(manifest, indent=2))
print(f'Written manifest to {manifest_path}')
```

```python
# Backup original files

from datetime import datetime
from pathlib import Path
import shutil

root = Path.cwd()
backup_dir = root.parent / f"incident2kb_backup_{datetime.utcnow():%Y%m%d_%H%M%S}"
shutil.copytree(root, backup_dir)
print(f"Backup created at {backup_dir}")
```
